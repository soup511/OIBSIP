```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

Load the Iris dataset

```python
from sklearn.datasets import load_iris
```

```python
iris = load_iris()
```

```python
iris
```

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)
```

```python
df
```

 Add target column

```python
df["species"] = iris.target

df.head(5)
```

Convert target numbers into species names

```python
df["species"] = df["species"].map({
    0: "Setosa",
    1: "Versicolor",
    2: "Virginica"
})

df.sample(5)
```

**Visualizations**

**Feature Selection Discussion**

```python
features = iris.feature_names
correlation_matrix = df[features].corr()

plt.figure(figsize=(8, 6))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Feature Correlation Matrix")
plt.show()
```

**Train/Test Split**

```python
from sklearn.model_selection import train_test_split
```

```python
X = df[iris.feature_names]
y = iris.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.20, random_state=42)
```

```python
print("Training samples:", X_train.shape[0])
print("Testing samples:", X_test.shape[0])
```

```python
from sklearn.linear_model import LogisticRegression
```

**Model 1 — Logistic Regression**

```python
logisticmodel = LogisticRegression(max_iter=200)
logisticmodel.fit(X_train, y_train)
```

```python
y_pred_logistic = logisticmodel.predict(X_test)
```

```python
train_accuracy = logisticmodel.score(X_train, y_train)
train_accuracy
```

```python
test_accuracy = logisticmodel.score(X_test, y_test)
test_accuracy
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test,y_pred_logistic  )
```

```python
from sklearn.metrics import classification_report
classification_report(y_test,y_pred_logistic  )
```

**Model 2 — K-Nearest Neighbours**

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn_model = KNeighborsClassifier(n_neighbors=5)
knn_model.fit(X_train, y_train)
```

```python
y_pred_knn = knn_model.predict(X_test)
```

```python
train_accuracy = knn_model.score(X_train, y_train)
train_accuracy
```

```python
test_accuracy = knn_model.score(X_test, y_test)
test_accuracy
```

```python
confusion_matrix(y_test,y_pred_knn)
```

```python
classification_report(y_test,y_pred_knn)
```

**Model 3-Decision Tree**

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(X_train, y_train)
```

```python
y_pred_dt=dt.predict(X_test)
```

```python
train_accuracy = dt.score(X_train, y_train)
train_accuracy
```

```python
test_accuracy = dt.score(X_test, y_test)
test_accuracy
```

```python
confusion_matrix(y_test,y_pred_dt)
```

```python
classification_report(y_test,y_pred_dt)
```

**Compare all 3 models**

```python
from sklearn.metrics import accuracy_score

models = {
    "Logistic Regression": (logisticmodel, y_pred_logistic),
    "K-Nearest Neighbours": (knn_model, y_pred_knn),
    "Decision Tree": (dt, y_pred_dt)
}

results = []

for name, (model, predictions) in models.items():
    results.append({
        "Model": name,
        "Train Accuracy": model.score(X_train, y_train),
        "Test Accuracy": accuracy_score(y_test, predictions)
    })

results_df = pd.DataFrame(results)

results_df
```

**Compare Precision, Recall and F1**

```python
metrics = []

for name, (model, predictions) in models.items():
    report = classification_report(
        y_test,
        predictions,
        output_dict=True
    )

    metrics.append({
        "Model": name,
        "Precision": report["weighted avg"]["precision"],
        "Recall": report["weighted avg"]["recall"],
        "F1-Score": report["weighted avg"]["f1-score"]
    })

metrics_df = pd.DataFrame(metrics)

metrics_df
```

**Show all confusion matric**

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

for ax, (name, (model, predictions)) in zip(axes, models.items()):
    cm = confusion_matrix(y_test, predictions)

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        ax=ax,
        xticklabels=iris.target_names,
        yticklabels=iris.target_names
    )

    ax.set_title(name)
    ax.set_xlabel("Predicted")
    ax.set_ylabel("Actual")

plt.tight_layout()
plt.show()
```

**Accuracy comparison graph**

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=results_df,
    x="Model",
    y="Test Accuracy"
)

plt.ylim(0, 1.05)
plt.title("Model Accuracy Comparison")
plt.xlabel("Model")
plt.ylabel("Test Accuracy")
plt.xticks(rotation=15)

plt.show()
```

**Identify the best model**

```python
best_model = results_df.loc[
    results_df["Test Accuracy"].idxmax()
]

print("Best Model:", best_model["Model"])
print("Test Accuracy:", best_model["Test Accuracy"])
```

## Best-Performing Model

Three classification models were evaluated: Logistic Regression, K-Nearest Neighbours, and Decision Tree.

All three models achieved 100% accuracy on the test dataset. Logistic Regression achieved 97.5% training accuracy, K-Nearest Neighbours achieved approximately 96.67% training accuracy, and Decision Tree achieved 100% training accuracy.

Since all three models achieved the same test accuracy, accuracy alone cannot distinguish the best model. Logistic Regression is selected as the preferred model because it provides excellent predictive performance while maintaining a relatively simple and interpretable structure. The Decision Tree achieved perfect training accuracy, which may indicate a greater potential for overfitting, although the test results do not show evidence of overfitting.

Therefore, **Logistic Regression is selected as the preferred model for this project.**
